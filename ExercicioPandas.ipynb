***Diagnóstico e limpeza***

```python
linhas, colunas = df_vendas.shape
print(f'dimensões: {linhas} linhas x {colunas} colunas')
```

```python
df_vendas.info()
```

```python
df_vendas['valor'] = df_vendas['valor'].fillna(df_vendas['valor'].median())
```

```python
df_vendas.loc[(df_vendas['valor'] > 500) & (df_vendas['status'] == 'Concluído'), ['cliente_id','valor','status']]
```

***Cruzamento e Transformação***

```python
pd.merge(df_vendas, df_clientes, on='cliente_id', how='left')
```

```python
df_media = pd.merge(df_vendas, df_clientes, on='cliente_id', how='left')
df_media['media_categoria'] = df_media.groupby('categoria')['valor'].transform('mean')
df_media.head()
```

```python
gmail = df_vendas[df_vendas['email'].str.contains('@gmail.com')]
print('Emails do Gmail: ', len(gmail))
```

***Analise Temporal e Agregação***

```python
df_media['data_hora'] = pd.to_datetime(df_media['data_hora'])
df_media['mês'] = df_media['data_hora'].dt.month
df_media['dia_semana'] = df_media['data_hora'].dt.day_name()
df_media[['data_hora','mês','dia_semana']].head()
```

```python
pd.pivot_table(df_media, values='valor', index='categoria', columns='cidade', aggfunc='sum', fill_value=0)
```
