**Reorganizando dados com `reshape()`**

```python
# Criar um array unidimensional com 12 elementos (representando visitas mensais)
visitas_mensais = np.array([10000, 11000, 12000, 10500, 11500, 12500, 13000, 12000, 11000, 10000, 9500, 9000])

print("Array unidimensional de visitas mensais:")
display(visitas_mensais)
```

```python
# Reorganizar o array em uma matriz 4x3 (4 trimestres com 3 meses cada)
matriz_trimestral_visitas = visitas_mensais.reshape(4, 3)

print("Matriz trimestral de visitas (4 trimestres, 3 meses/trimestre):")
display(matriz_trimestral_visitas)
```

**Combinando Matrizes com `np.vstack()`**

```python
# Dados de visitas para o primeiro semestre (exemplo: 2 trimestres x 3 meses cada)
visitas_primeiro_semestre = np.array([
    [10000, 11000, 12000],  # 1º Trimestre
    [10500, 11500, 12500]   # 2º Trimestre
])

print("Dados de visitas do Primeiro Semestre:")
display(visitas_primeiro_semestre)

# Dados de visitas para o segundo semestre (exemplo: 2 trimestres x 3 meses cada)
visitas_segundo_semestre = np.array([
    [13000, 12000, 11000],  # 3º Trimestre
    [10000, 9500, 9000]    # 4º Trimestre
])

print("\nDados de visitas do Segundo Semestre:")
display(visitas_segundo_semestre)
```

```python
# Combinar as duas matrizes verticalmente usando np.vstack()
visitas_anuais_combinadas = np.vstack((visitas_primeiro_semestre, visitas_segundo_semestre))

print("\nMatriz combinada de visitas anuais:")
display(visitas_anuais_combinadas)
```

**Extraindo o Primeiro Trimestre com `copy()` para Projeção**

```python
# Extrair o primeiro trimestre (primeira linha) da matriz combinada e criar uma cópia
primeiro_trimestre_projecao = visitas_anuais_combinadas[0, :].copy()

print("Primeiro Trimestre para Projeção (cópia):")
display(primeiro_trimestre_projecao)

# Exemplo para mostrar que a cópia é independente:
# Alterando um valor na cópia não afeta o array original
primeiro_trimestre_projecao[0] = 99999
print("\nPrimeiro Trimestre de Projeção após alteração:")
display(primeiro_trimestre_projecao)

print("\nVerificando o array original (deve permanecer inalterado):")
display(visitas_anuais_combinadas)
```

**Exportando Dados para Relatório com `ravel()`**

```python
# Achatar a matriz de visitas anuais combinadas em um array 1D
relatorio_visitas_1d = visitas_anuais_combinadas.ravel()

print("Array 1D de visitas para relatório:")
display(relatorio_visitas_1d)
```
